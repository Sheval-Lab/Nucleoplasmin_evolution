```python
#read fasta file

npm_fasta = "/home/ilnitsky/NPM/orthodb_findings/Metazoa_NPM_cdhit/mmseqs-easycluster_all_seqs_no_cluster_names.fasta"
ncl_fasta = "/home/ilnitsky/NPM/orthodb_findings/Metazoa_NCL_mmseqs_easycluster/ncl_mmseqs_easycluster_all_seqs_no_cluster_names.fasta"
fbl_fasta = "/home/ilnitsky/NPM/orthodb_findings/Metazoa_FBL_mmseqs_easycluster/fbl_mmseqs_easycluster_all_seqs_no_cluster_names.fasta"
tco_fasta = "/home/ilnitsky/NPM/orthodb_findings/Metazoa_TCOF1_mmseqs_easycluster/tco_mmseqs_easycluster_all_seqs_no_cluster_names.fasta"

def fasta_to_dataframe(file_path):
    sequences = []
    names = []
    with open(file_path, "r") as f:
        lines = f.readlines()
        for line in lines:
            if line.startswith(">"):
                names.append(line.strip().lstrip(">"))
            else:
                sequences.append(line.strip())
    df = pd.DataFrame({"name": names, "sequence": sequences})
    return df

npm_df = fasta_to_dataframe(npm_fasta)
ncl_df = fasta_to_dataframe(ncl_fasta)
fbl_df = fasta_to_dataframe(fbl_fasta)
tco_df = fasta_to_dataframe(tco_fasta)

npm_df
```

```python
import sys
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from localcider.sequenceParameters import SequenceParameters
from ete3 import NCBITaxa
import random
sys.path.append('/home/ilnitsky/tools/AIUPred')
import aiupred_lib

# Assuming your dataframe is called 'df' and has the columns:
# 'species', 'sequence_npm', 'sequence_ncl', 'sequence', 'sequence_tco'

# Get all species names
df = sorted_presence_df
species_names = df['species'].tolist()

# Create figure with 4 columns (NPM, NCL, Sequence, TCO)
fig, axs = plt.subplots(len(species_names), 4, figsize=(20, 1*len(species_names)), squeeze=False)

for i, species_name in enumerate(species_names):
    row = df[df['species'] == species_name].iloc[0]
    
    # Process each protein type (NPM, NCL, Sequence, TCO)
    protein_columns = ['sequence_npm', 'sequence_ncl', 'sequence', 'sequence_tco']
    protein_labels = ['NPM', 'NCL', 'Sequence', 'TCO']
    
    for j, (protein_col, protein_label) in enumerate(zip(protein_columns, protein_labels)):
        ax1 = axs[i, j]
        ax2 = ax1.twinx()
        
        sequence = row[protein_col]
        
        if pd.notna(sequence) and len(str(sequence)) > 0:
            # AIUPred disorder prediction
            disorder_prediction = aiupred_lib.aiupred_disorder(sequence)
            positions = range(1, len(sequence) + 1)

            # LocalCIDER NCPR data
            try:
                SeqOb = SequenceParameters(sequence)
            except:
                print(f"Error with sequence {i}: {species_name} {protein_label}")
                continue
            ncpr_data = SeqOb.get_linear_NCPR(blobLen=5)
            ncpr_x = ncpr_data[0, :]
            ncpr_y = ncpr_data[1, :]

            # Plot disorder score
            ax1.plot(positions, disorder_prediction, 'b-', linewidth=2, label='Disorder Score')
            ax1.set_ylim([0, 1.05])
            ax1.tick_params(axis='both', which='both', length=0, labelleft=False, labelbottom=False)
            ax1.grid(True, alpha=0.3)

            # Plot NCPR
            for k in range(len(ncpr_x)):
                color = 'green' if ncpr_y[k] >= 0 else 'red'
                ax2.bar(ncpr_x[k], ncpr_y[k], width=1, color=color, alpha=0.5, align='center')

            # Set NCPR axis properties
            ax2.set_ylim([-1.0, 1.0])
            ax2.tick_params(axis='both', which='both', length=0, labelleft=False, labelbottom=False)

        # Set title for each subplot
        # ax1.set_title(f'{species_name} - {protein_label}', fontsize=8)
        ax2.set_xticks([])
        ax2.set_yticks([])
        ax2.set_xlabel('')
        ax2.set_ylabel('')
        ax1.set_xlabel('')
        ax1.set_ylabel('')
        ax1.set_xticks([])
        ax1.set_yticks([])
        # Add legend (only for first valid plot)
        if i == 0 and j == 0 and pd.notna(sequence) and len(str(sequence)) > 0:
            blue_patch = plt.Line2D([0], [0], color='blue', linewidth=2, label='Disorder Score')
            green_patch = plt.Rectangle((0,0), 1, 1, color='green', alpha=0.5, label='Positive NCPR')
            red_patch = plt.Rectangle((0,0), 1, 1, color='red', alpha=0.5, label='Negative NCPR')
            # Uncomment the next line if you want to show the legend
            # ax1.legend(handles=[blue_patch, green_patch, red_patch], loc='upper right')

plt.tight_layout()
plt.gcf().set_size_inches(20, 1*len(species_names))
plt.savefig("/home/ilnitsky/NPM/orthodb_findings/Metazoa_NPM_NCL_FBL_TCOF_aiupred_cider_plot.png")
plt.savefig("/home/ilnitsky/NPM/orthodb_findings/Metazoa_NPM_NCL_FBL_TCOF_aiupred_cider_plot.pdf") 
```
